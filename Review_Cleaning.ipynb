```python
# Install libraries if you haven't already (uncomment the line below if needed)
# !pip install pandas nltk

import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

# Download required NLTK data
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
# 1. Load the dataset
df = pd.read_csv('IMDB_Dataset_CLEANED.csv')

# 2. Inspect missing values and duplicate reviews
print("Missing Values per column:\n", df.isnull().sum())
print("\nTotal Duplicate Reviews:", df.duplicated().sum())

# Drop duplicates and missing values to keep the dataset clean
df.drop_duplicates(inplace=True)
df.dropna(inplace=True)

print(f"\nDataset shape after removing duplicates: {df.shape}")
```

```python
def clean_text(text):
    """Applies lowercase, regex cleaning, tokenization, stopword removal, and lemmatization."""
    
    # 3. Convert text to lowercase
    text = text.lower()
    
    # 4. Remove HTML tags, URLs, punctuation, numbers, and special characters
    text = re.sub(r'<.*?>', ' ', text) # Remove HTML tags (like )
    text = re.sub(r'http\S+|www\S+|https\S+', '', text, flags=re.MULTILINE) # Remove URLs
    text = re.sub(r'[^a-z\s]', '', text) # Keep only lowercase letters and spaces
    
    # 6. Tokenize the reviews (Split text into words)
    tokens = word_tokenize(text)
    
    # 5. Remove stopwords using NLTK
    stop_words = set(stopwords.words('english'))
    tokens = [word for word in tokens if word not in stop_words]
    
    # 7. Apply Lemmatization
    lemmatizer = WordNetLemmatizer()
    tokens = [lemmatizer.lemmatize(word) for word in tokens]
    
    # Join tokens back into a single string
    return ' '.join(tokens)
```

```python
# 8. Store the cleaned reviews in a new column
print("Cleaning in progress... This may take a couple of minutes.")
df['cleaned_review'] = df['review'].apply(clean_text)
print("Cleaning complete!")
```

```python
# 9. Compare original and cleaned reviews
# Set pandas option to show more text
pd.set_option('display.max_colwidth', 150)
comparison_df = df[['review', 'cleaned_review']].head(3)
display(comparison_df)

# 10. Export the preprocessed dataset as CSV
df.to_csv('cleaned_imdb_reviews.csv', index=False)
print("\nDataset successfully exported as 'cleaned_imdb_reviews.csv'")
```
