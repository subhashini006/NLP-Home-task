```python
import re
from collections import Counter, defaultdict
from datasets import load_dataset
```

```python
dataset = load_dataset(
    "Salesforce/wikitext",
    "wikitext-2-raw-v1"
)

print(dataset)
```

```python
train_data = dataset["train"]

text = " ".join(train_data["text"])

print("Total characters:", len(text))
print(text[:1000])
```

```python
def clean_text(text):
    text = text.lower()
    
    # Remove unwanted characters
    text = re.sub(r"[^a-zA-Z0-9\s]", " ", text)
    
    # Remove extra spaces
    text = re.sub(r"\s+", " ", text)
    
    return text.strip()
```

```python
cleaned_text = clean_text(text)

print(cleaned_text[:500])
```

```python
tokens = cleaned_text.split()

print("Number of tokens:", len(tokens))
print(tokens[:50])
```

```python
unigram = Counter(tokens)

print("Most common words:")
print(unigram.most_common(20))
```

```python
bigram = Counter(
    zip(tokens, tokens[1:])
)

print("Most common bigrams:")
print(bigram.most_common(20))
```

```python
trigram = Counter(
    zip(tokens, tokens[1:], tokens[2:])
)

print("Most common trigrams:")
print(trigram.most_common(20))
```

```python
def bigram_probability(word1, word2):
    count_bigram = bigram[(word1, word2)]
    count_word = unigram[word1]
    
    if count_word == 0:
        return 0
    
    return count_bigram / count_word
```

```python
def trigram_probability(word1, word2, word3):
    count_trigram = trigram[(word1, word2, word3)]
    count_bigram = bigram[(word1, word2)]
    
    if count_bigram == 0:
        return 0
    
    return count_trigram / count_bigram
```

```python
def predict_next_words(sentence, top_n=5):
    
    sentence = clean_text(sentence)
    words = sentence.split()
    
    if len(words) == 0:
        return []
    
    predictions = []
    
    # Use trigram if at least two words are available
    if len(words) >= 2:
        
        word1 = words[-2]
        word2 = words[-1]
        
        for (w1, w2, w3), count in trigram.items():
            
            if w1 == word1 and w2 == word2:
                
                probability = count / bigram[(w1, w2)]
                
                predictions.append(
                    (w3, probability)
                )
    
    # If no trigram prediction, use bigram
    if len(predictions) == 0:
        
        word = words[-1]
        
        for (w1, w2), count in bigram.items():
            
            if w1 == word:
                
                probability = count / unigram[word]
                
                predictions.append(
                    (w2, probability)
                )
    
    # Sort by probability
    predictions.sort(
        key=lambda x: x[1],
        reverse=True
    )
    
    return predictions[:top_n]
```

```python
sentence = "machine learning"

predictions = predict_next_words(
    sentence,
    top_n=5
)

print("Input:", sentence)
print("\nTop predicted next words:")

for i, (word, probability) in enumerate(predictions, 1):
    print(
        f"{i}. {word} - {probability:.3f}"
    )
```

```python
while True:
    
    sentence = input("\nEnter a sentence: ")
    
    if sentence.lower() == "exit":
        print("Predictor stopped.")
        break
    
    predictions = predict_next_words(
        sentence,
        top_n=5
    )
    
    print("\nTop predicted next words:")
    
    if predictions:
        
        for i, (word, probability) in enumerate(predictions, 1):
            print(
                f"{i}. {word} - {probability:.3f}"
            )
    
    else:
        print("No prediction found.")
```
